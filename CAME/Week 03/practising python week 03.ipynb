### **Exception Handling**

```python
raw_values = ['0.91','bad',' ','0.72','0.65','']
clean_values=[]
for index in range(len(raw_values)):
    try:
        raw_values[index]=float(raw_values[index])
        clean_values.append(raw_values[index])
    except ValueError:
        raw_values[index]=raw_values[index].strip()
print(raw_values)
print(clean_values if len(clean_values)!=0 else "No valid values")
```

```python
predictions = ['cat','dog','cat','bird','dog']
true_labels = ['cat','cat','cat','bird','cat']
mistakes = []
for index,predict in enumerate(predictions):
    if predict == true_labels[index]:
        pass
    else:
        mistakes.append((index,predict,true_labels[index]))
print(f'mistake count: {len(mistakes)} & mistake with highest index: {mistakes[-1][0]}')
```

```python
results = [('cat',0.97),('dog',0.62),('bird',0.88),('cat',0.75)]
highest_confidence=0
highest_label=''
for label,confidence in results:
    if confidence >= 0.85:
        print(label,confidence,sep=' -> ')
        if confidence>highest_confidence:
            highest_confidence,highest_label=confidence,label
print((highest_label,highest_confidence))
```
