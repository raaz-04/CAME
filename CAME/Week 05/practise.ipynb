>>> ### ***Inheritance***

```python
class vehicle:
    has_wheels = True

    def display(self):
        print("This is a vehicle")

class car(vehicle):
    def display_info(self):
        print("This is a car")

car_01 = car()
print(car_01.has_wheels)
car_01.display()
car_01.display_info()
```

>>> ### ***Method Overriding***

```python
class vehicle:
    has_wheels = True

    def display(self):
        print("This is a vehicle")

class car(vehicle):
    def display(self):
        print("This is a car")

car_01 = car()
print(car_01.has_wheels)
car_01.display()
```

```python
class vehicle:
    has_wheels = True

    def __init__(self,manufacturing_date,color):
        self.manufacturing_date = manufacturing_date
        self.color = color

    def display(self):
        print("This is a vehicle")

class car(vehicle):
    def __init__(self, manufacturing_date, color,brand):
        super().__init__(manufacturing_date, color)
        self.brand = brand

    def display(self):
        print(f"manufacturing date : {self.manufacturing_date} | color : {self.color} | brand : {self.brand}")

car_01 = car("2026-06-28","red","Toyota")
car_01.display()
```

>>> ### ***Encapsulation***

<#>  _(method name) / _(attribute name) ==> protected

```python
class car:
    wheel_count = 4

    def __init__(self, brand, color):
        self._color = color #protected attributes
        self._brand = brand

    def _display(self):  #protected method
        print(f"{self._brand} , {self._color} color car")

car_01 =car('BMW', 'red')
print(car_01._color,car_01._brand)
car_01._display()
```

No Compile errors.In python protected attributes impliy the programmer to do not access these attributes outside the class.This is a bad practise.<br>Only recommended the access through sub classes.


```python
class vehicle:
    wheel_count = 4

    def __init__(self, brand, color):
        self._color = color #protected attributes
        self._brand = brand

    def _display(self):  #protected method
        print(f"{self._brand} , {self._color} color car")

class car(vehicle):
    def __init__(self, brand, color, manufacturing_date):
        super().__init__(brand, color)
        self.manufacturing_date = manufacturing_date

    def _display(self):
        print(f"manufacturing date : {self.manufacturing_date} | brand : {self._brand} | color : {self._color}")

car_01 =car('BMW', 'red', '2026-06-28')
car_01._display()
```

<#>  _ _(method name) / _ _(attribute name) ==> private

```python
class car:
    wheel_count = 4

    def __init__(self, brand, color):
        self.__color = color #private attributes
        self.__brand = brand

    def __display(self):  #private method
        print(f"{self.__brand} , {self.__color} color car")

car_01 =car('BMW', 'red')
print(car_01.__color,car_01.__brand)
car_01.__display()
```

Private attributes/methods can not access through the outside of the class. 

```python
class vehicle:
    wheel_count = 4

    def __init__(self, brand, color):
        self.__color = color #protected attributes
        self.__brand = brand

    def __display(self):  #protected method
        print(f"{self.__brand} , {self.__color} color car")

class car(vehicle):
    def __init__(self, brand, color, manufacturing_date):
        super().__init__(brand, color)
        self.manufacturing_date = manufacturing_date

    def display(self):
        print(f"manufacturing date : {self.manufacturing_date} | brand : {self.__brand} | color : {self.__color}")

car_01 =car('BMW', 'red', '2026-06-28')
car_01.display()
```

### ***@Property* Decorator** 

This makes a method acts like an attribute.<br>No need of adding parenthesis when calling the method.

```python
class car:
    wheel_count = 4

    def __init__(self, brand, color):
        self.__color = color #private attributes
        self.__brand = brand

    def __display(self):  #private method
        print(f"{self.__brand} , {self.__color} color car")

    @property
    def display_info(self):
        self.__display()

car_01 =car('BMW', 'red')
car_01.display_info
```

>>> ### ***Composition***

Inheritance => "IS-A" relationship<br>
Composition => "HAS-A" relationship

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f'Engine with {self.horsepower} HP is roaring to life!'
    
class Battery:
    def __init__(self, capacity):
        self.capacity = capacity

    def charge(self):
        return f"Battery charging to {self.capacity} kWh..."
    
class Car:
    def __init__(self, model, horsepower, charging_capacity):
        self.model = model
        self.horsepower = horsepower
        self.capacity = charging_capacity
        self.engine = Engine(horsepower)
        self.battery = Battery(charging_capacity)
car_01 = Car('Toyota', 220, 1)
print(car_01.engine.start())
print(car_01.battery.charge())
```

### Mathematics

```python
def vector_addition(vector_01,vector_02):
    #return [vector_01[index]+vector_02[index] for index in range(len(vector_01))]
    return list(map(lambda x,y:x+y , vector_01,vector_02))

customer_01 = [25, 5000, 3]
customer_02 = [21, 7000, 9]
print(vector_addition(customer_01,customer_02))

```

```python
def scalar_multiplication(vector_01,scalar_multiplier):
    return list(map(lambda x:x*scalar_multiplier , vector_01))

customer_01 = [25, 5000, 3]
scalar_constant = 3
print(scalar_multiplication(customer_01,scalar_constant))
```

### <u>Magnitude of a vector</u>

a = (a1, a2)<br>
|a| = [a1^2 + a2^2]**0.5

#### Q. Find magnitude of the vector **a** = (1,-2,2)<br>|**a**| = [1+4+4]**0.5 = 3

```python
from math import sqrt as sqrt
def vector_magnitude(vector):
    return sqrt(sum(list(map(lambda x:x**2 , vector))))

vector_01 = [25, 5000, 3]
print(vector_magnitude(vector_01))
print(vector_magnitude([1,-2,2]))
```
