### Default and Keyword Arguments

```python
def calculate_accuracy(correct_labels,total_labels=200):
    return f"{correct_labels/total_labels:.2%}"

print(calculate_accuracy(80)) #Default argument
print(calculate_accuracy(80,total_labels=100)) #Keyword argument
```

### *args and **kwargs

```python
def mean(*args):
    if not args:
        return 0.0
    print(f'Argument passed : {args}')
    return sum(args)/len(args)
print(f'Mean : {mean(1,4,8)}')

def average(*args):
    if not args:
        return "Not a validated argument"
    return sum(args)/len(args)
print(f'Average : {average(45,90,78,83)}')
```

```python
def average(**kwargs):
        if kwargs != None:
            print(f'Keyword Argument Passed : {kwargs}')
            return sum(kwargs.values())/len(kwargs.values())
        else:
            return 'Not a validated argument'
print(f'Average : {average(maths=80,science=78,history=93,sinhala=82):.2f}')
```

```python
def calculate_area(radius):
    """
    Calculate the area of the circle given its radius.
    Args:
        radius(float):The radius of the circle.
    Returns:
        float:The calculated area.
    """
    from math import pi as pi
    return pi*(radius**2)
print(calculate_area(5))

#We can get the docstring as below.
print(calculate_area.__doc__)

#-------------------------------
help(calculate_area)
```

### **A. sorted()** <br>
#### sorted(iterable,key,reverse=False)<br>


```python
my_list = [19,23,45,77,98,3]

print(sorted(my_list,reverse=True))
print(sorted(my_list,key=lambda x:x%3,reverse=False))

my_list.sort()
print(my_list)
```

```python
students = [('John',85),('Alice',92),('Bob',78)]
print(f'sorted list according to marks : {sorted(students,key=lambda x:x[1])}')
print(f'sorted list according to names : {sorted(students,key=lambda x:x[0])}')
students.sort()
print(f'sorted list after sort() : {students}')
```

```python
students = [('John',85),('Alice',92),('Bob',78)]
print(list(reversed(students)))
print(f'reversed list after reversed() : {students}')
print(students)
students.reverse()
print(students)
```

```python
users = [{'name':'Alice','age':30},{'name':'Bob','age':25},{'name':'Charlie','age':35}]
sort_users = sorted(users,key = lambda x : x['age'])
print(f'sorted users according to the age : {sort_users}')
```

### **B. min() & max()** <br>
#### min(iterable,key)  &  max(iterable,key)<br>

```python
students = [('John',85),('Alice',92),('Bob',78)]
print(f'minimum according to marks : {min(students,key=lambda x:x[1])}')
print(f'minimum according to length of name : {min(students,key=lambda x:len(x[0]))}')

print(f'maximum according to marks : {max(students,key=lambda x:x[1])}')
print(f'maximum according to length of name : {max(students,key=lambda x:len(x[0]))}')
```

### **C. map()** <br>
#### map(expression,iterable)

```python
users = [{'name':'Alice','age':30},{'name':'Bob','age':25},{'name':'Charlie','age':35}]
incrmental_age = 20
print(f'mapping users age after {incrmental_age} years : {list(map(lambda x : (x['name'],x['age']+incrmental_age),users))}')

subject_marks = {'maths': 80, 'science': 78, 'history': 93, 'sinhala': 82}
extra_mark = 3
print(f'after adding extra marks : {dict(map(lambda x:(x[0],x[1]+extra_mark),subject_marks.items()))}')
```

### **D. filter()** <br>
#### filter(expression,iterable)

```python
users = [{'name':'Alice','age':30},{'name':'Bob','age':25},{'name':'Charlie','age':35}]
threshold_age = 27
print(f'filtering users who are above threshold age {incrmental_age} years : {list(filter(lambda x :x['age']>threshold_age,users))}')

subject_marks = {'maths': 80, 'science': 78, 'history': 93, 'sinhala': 82}
pass_mark = 80
print(f'pass subjects : {dict(filter(lambda x:x[1]>=pass_mark,subject_marks.items()))}')
```

### **List Comprehension**

```python
my_list = [8,5,9,4,20]
squared_list = [item **2 for item in my_list]
print(f'squared list : {squared_list}')

new_list = [item**2 for item in my_list if item>5]
print(f'squared list if item > 5 : {new_list}')
```

```python
labels = ['cat','dog','fish','bear']
print(f'after mapping to dictionary : {dict((x[1],x[0]) for x in enumerate(labels))}')

print(f'after mapping to dictionary : {dict((y,x) for x,y in enumerate(labels))}')

print({y:x for x,y in enumerate(labels)})

labels = ['cat','dog','fish','bear']
print(dict(map(lambda x:(x[1],x[0]),enumerate(labels))))
```

### ***OOP (Object Oriented Programming)***

class House: (You have to follow pascal case when naming class names) Pascal case:Capitalize()<br>
    has_a_roof = 'True' (class attributes)<br>
>>> class attributes are shared between every object of the class.

House_01 = House()<br>
House_02 = House()<br>

```python
class House:
    has_a_roof = 'True'

    def repaint(self):  #first parameter is the object its self.(not a keyword/implies parameterless)
        print('House needs repainting')

house_01 = House()
house_02 = House()
print(house_01.has_a_roof,house_02.has_a_roof) #class attributes are shared between class objects.
house_01.repaint()
```

Write a python class to represent a circle.Define a method to calculate the area of a circle.Show the functionality creating an instance of the class.<br>

```python
from math import pi as pi 
class Circle:
    def calculate_area(self,radius):
        return pi*radius**2
    
circle_01 = Circle()
print(circle_01.calculate_area(5))
```

### **<u>Instance attributes and Constructor</u>** ***( _ _ init _ _ )***

```python
from math import pi as pi 
class Circle:
    def __init__(self,radius):  #constructor
        self.radius = radius   #instance attributes

    def calculate_area(self):
        return pi*self.radius**2
    
circle_01 = Circle(5)
print(circle_01.calculate_area())
```
