### <u>***<div>DataFrame NaN containing column Filling</div>***</u>

```python
import pandas as pd
df = pd.read_csv('customers_tiny.csv')
age_mean = df['age'].mean()
score_median = df['score'].median()
# df['age'] , df['score'] = df['age'].fillna(age_mean) , df['score'].fillna(score_median)
df = df.fillna({'age':age_mean,'score':score_median})
print(df)
```

```python
import pandas as pd
df = pd.read_csv('customers_tiny.csv')
df = df.dropna()
print(df)
df['age'] = df['age'].astype('int64')
print('----------------------------------')
print(df)
print('age dtype :',df['age'].dtype)
```

```python
import pandas as pd
df = pd.read_csv('customers_tiny.csv')
age_mean = df['age'].mean()
score_median = df['score'].median()
df = df.fillna({'age':age_mean,'score':score_median})
df['age']  = df['age'].astype('int32')
df = df.sort_values('score')
print(df)
print('---------------------------------')
df.sort_values('age',ascending=False)
```

### <u>***<div>DataFrame groupby() Command</div>***</u>

```python
import pandas as pd
df = pd.read_csv('customers_tiny.csv')
df = df.dropna()
print(df.groupby('city')['score'].mean())
```

```python
import pandas as pd
df = pd.read_csv('customers_tiny.csv')
df = df.dropna()
new_df = df.groupby('city').agg(
    {'score': 'mean','age': 'mean'}
)
print(new_df)
print('--------------------')
df.groupby('city').agg(
    avg_score  = ('score', 'mean'),
    n_rows = ('score', 'count'),
    avg_age = ('age', 'mean'),
)
```
