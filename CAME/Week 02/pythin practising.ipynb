#### ***>> Mutability***

#### ***<u>Lists Mutability</u>***

```python
sample_list = [10,6,8,6]

print(f"List before updating :{sample_list}")
sample_list[0] = 5
print(f"List after updating : {sample_list}")
sample_list.insert(0,9)
print(f"List after inserting :{sample_list}")
sample_list.append(101)
print(f"List after appending :{sample_list}")
sample_list.extend([10,1,2,3])
print(f"List after extending :{sample_list}")
sample_list.remove(101)
print(f"List after removing :{sample_list}")
x = sample_list.pop(3)
print(f"List after popping :{sample_list} & Popped item : {x}")
```

#### ***<u>Tuple is Immutable</u>***

```python
sample_tuple = (8,4,3,45,3,87,3,0,3)

# print(f'Tuple if we try to mutate : {sample_tuple.remove(8)}')
# print(f'Tuple if we try to mutate : {sample_tuple.append(8)}')
''' ---------------------------------------------------------------------------
AttributeError                            Traceback (most recent call last)
Cell In[15], line 3
      1 sample_tuple = (8,4,3)
      2 
----> 3 print(f'Tuple if we try to mutate : {sample_tuple.remove(8)}')

AttributeError: 'tuple' object has no attribute 'remove'.
---------------------------------------------------------------------------'''

print(f"Number of 3's in tuple : {sample_tuple.count(3)}")
print(f"45's index in tuple : {sample_tuple.index(45)}")

print(f"Tuple before mutating {sample_tuple}")
temp = list(sample_tuple)
temp.insert(3,200)
print(f"Tuple after mutating {tuple(temp)}")

sample_tuple += (3,2,3)
print(sample_tuple)
```

#### ***<u>Sets Mutability</u>***

```python
sample_set = {6,3,5,1,8,11}
print(f"Set before mutating : {sample_set}")
sample_set.add(101)
print(f"Set after adding : {sample_set}")
sample_set.update({101,202,303})
print(f"Set after updating : {sample_set}")
sample_set.remove(101)
print(f"Set after removing : {sample_set}") #Error if there is no such element
sample_set.discard(1)
print(f"Set after discarding : {sample_set}")  #No error even if there is no such element
```

#### ***<u>Sets calculations</u>***

```python
set_A = {10,12,5,3,8}
set_B = {8,15,20,18}

print(f"A U B : {set_A.union(set_B)}")
print(f"A U B : {set_A | set_B}")
print(f"A intersect B : {set_A.intersection(set_B)}")
print(f"A intersect B : {set_A & set_B}")
print(f"A - B : {set_A.difference(set_B)}")
print(f"B - A : {set_B.difference(set_A)}")
```
