### **01. _ _ len _ _()**

```python
class shoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = ['laptop', 'mouse', 'pendrive']

    def __len__(self):
        return len(self.items)
    

myCart = shoppingCart('Rasula')
len(myCart)
```

### **02. _ _ getitem _ _()**

```python
class shoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = ['laptop', 'mouse', 'pendrive']

    def __len__(self):
        return len(self.items)
    
    def __getitem__(self, index):
        return self.items[index]
    

myCart = shoppingCart('Rasula')
myCart[1]
```

```python
class SimpleDataset:
    def __init__(self, samples, labels):
        self.samples = samples
        self.labels = labels

    def __len__(self):
        return len(self.samples)
    
    def __getitem__(self, index):
        return self.samples[index], self.labels[index]
    
data = SimpleDataset(['a', 'b', 'c'], [0, 1, 0])
print(len(data))
print(data[1])

for x,y in data:
    print(x, y)
```

### **03. _ _ call _ _()**

```python
class scalar:
    def __init__(self, factor):
        self.factor = factor
    
    def __call__(self, x):
        return self.factor*x

my_scalar = scalar(2)
my_scalar(10)
```

### **04. _ _ repr _ _()**

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f'person(name={self.name}, age={self.age})'

person_01 = Person('John', 28)
print(person_01)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f'person(name={self.name}, age={self.age})'
    
    def __str__(self):
        return 'This is the person class'

person_01 = Person('John', 28)
print(person_01)
print(repr(person_01)) #person_01.__repr__()
```

__Q. Find the distance between vectors,__<br>
__a__ = (1,-1,3) & __b__ = (3,2,-2)

|__a-b__| = [(3-1)^2+(2+1)^2+(3+2)^2]^0.5
      = root 38

```python
import math
def distance_between_vectors(v1, v2):
    if len(v1)==len(v2):
        try:
            return math.sqrt(sum(list(map(lambda x,y: pow((x-y),2),v1,v2))))
        except TypeError:
            return f"Invalid vectors"
    raise ValueError('Dimensions are different')

print(distance_between_vectors([1,-1,3],[3,2,-2]))
print(distance_between_vectors([1,-1,'r'],[3,2,-2]))
print(distance_between_vectors([1,2],[1,3,5]))
```

__Q. In a movie recommendation system,every movie has been rated on a scale from 1 to 10<br>based on two features which are action level and comedy level.__<br>
__movie_A = [1,2] , movie_B = [8,8] , user_reference = [2,2]__<br>
I. Normalize all three vectors by dividing magnitudes<br>
II. Mark all newly obtain vectors in a coordinate plane<br>
III. Calculate the eucliedian distance from user reference vector<br>
IV. Based on the above calculatiuons which movie do you suggest to the user? 

```python
import math
def normalize_vectors(v):
    return [x/math.sqrt(sum([x**2 for x in v])) for x in v]

def distance_between_vectors(v1, v2):
    if len(v1)==len(v2):
        try:
            return math.sqrt(sum(list(map(lambda x,y: pow((x-y),2),v1,v2))))
        except TypeError:
            return f"Invalid vectors"
    raise ValueError('Dimensions are different')

movie_A = [1, 2]
movie_B = [8, 8]
user_reference = [2,2]
movie_dict = {'movie_A':movie_A, 'movie_B':movie_B}
print(f'Normalized movie_A : {normalize_vectors(movie_A)}',f'Normalized movie_B : {normalize_vectors(movie_B)}',f'Normalized user_reference : {normalize_vectors(user_reference)}',sep='\n')
print('Eucliedian distance without normalizing :',list(map(lambda x:distance_between_vectors(x,user_reference),movie_dict.values())))
print('Based on only Eucliedian Distance :',[i[0] for i in movie_dict.items() if i[1]==min(movie_dict.values(),key=lambda x:distance_between_vectors(x,user_reference))])
movie_dict = {'movie_A':normalize_vectors(movie_A), 'movie_B':normalize_vectors(movie_B)}
print('Eucliedian distance with normalizing :',list(map(lambda x:distance_between_vectors(x,normalize_vectors(user_reference)),movie_dict.values())))
print('Based on Normalized Eucliedian Distance :',[i[0] for i in movie_dict.items() if i[1]==min(movie_dict.values(),key=lambda x:distance_between_vectors(x,normalize_vectors(user_reference)))])
```

>>> ## <b><u>Manhattan Distance</u> (L1 Norm)</b>

<table><td style='padding:20px'>Eucliedian Distance = Diagonal Distance<br>
Manhattan Distance = Horizontal Distance + Vertical Distance</td></table>

Q. Find L1 Norm between __v1__=(-1,3) & __v2__=(-4,-1)<br>
Manhattan distance = |-4+1| +|-1-3| = 7 units

```python
def Manhattan_distance(v1,v2):
    if len(v1)==len(v2):
        # return sum([abs(v1[i]-v2[i]) for i in range(len(v1))])
        return sum([abs(x-y) for x,y in zip(v1,v2)])
    raise ValueError('Dimensions should be same')
vector_01 = [-1,3]
vector_02 = [-4,-1]
vector_03 = [1,2,-3]
print(Manhattan_distance(vector_01,vector_02))
print(Manhattan_distance(vector_01,vector_03))
```

>>> ## <u><b>Feature Scaling</b></u>

If Some features have large numerical values,even though they have less<br>
significance they will dominate.So that is why we do the __scaling__

```python
def min_max_scale(value,min,max):
    return (value-min)/(max-min)

customer_a = [25,50000]
customer_b = [26,51000]
customer_c = [35,45000]
customer_d = [43,78000]
customer_e = [40,54000]
min_val1 = min([customer_a,customer_b,customer_c,customer_d,customer_e],key=lambda x:x[0])[0]
min_val2 = min([customer_a,customer_b,customer_c,customer_d,customer_e],key=lambda x:x[1])[1]
max_val1 = max([customer_a,customer_b,customer_c,customer_d,customer_e],key=lambda x:x[0])[0]
max_val2 = max([customer_a,customer_b,customer_c,customer_d,customer_e],key=lambda x:x[1])[1]
print([min_max_scale(customer_a[0],min_val1,max_val1),min_max_scale(customer_a[1],min_val2,max_val2)])
print([min_max_scale(customer_b[0],min_val1,max_val1),min_max_scale(customer_b[1],min_val2,max_val2)])
print([min_max_scale(customer_c[0],min_val1,max_val1),min_max_scale(customer_c[1],min_val2,max_val2)])
print([min_max_scale(customer_d[0],min_val1,max_val1),min_max_scale(customer_d[1],min_val2,max_val2)])
print([min_max_scale(customer_e[0],min_val1,max_val1),min_max_scale(customer_e[1],min_val2,max_val2)])
```
