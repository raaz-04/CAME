### Dot Product from Numpy Arrays

```python
list_01 = [1,2,3,4]
list_02 = [5,6,7,8]
array_01 = np.array(list_01)
array_02 = np.array(list_02)
print(sum(array_01*array_02))
###########################################
print('-------------From function---------')
def calcualte_dot_product(vector_01,vector_02):
    if len(vector_01)!=len(vector_02):
        return f'Dimensions are different'
    return sum(vector_01*vector_02)
print(calcualte_dot_product(array_01,array_02))
```
