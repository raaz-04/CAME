```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    'age' : [22, 23, 28, 41, 33, 26, 45, 31],
    'score' : [0.81, 0.44, 0.92, 0.67, 0.70, 0.55, 0.60, 0.80],
    'city' : ['Colombo', 'Kandy', 'Galle', 'Colombo', 'Galle', 'Kandy', 'Colombo', 'Galle']
})

#Histogram

plt.figure()
plt.hist(df['age'], bins=5, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('age')
plt.ylabel('count')
plt.show()
```

```python
city_mean = df.groupby('city')['score'].mean()
print(city_mean)
```

```python
#barchart
plt.figure()
city_mean.plot(kind='bar', color='orange', edgecolor='black')
plt.title('Average Score by City')
plt.xlabel('City')
plt.ylabel('Average Score')
plt.show()
```

<div style='color:red'>Why use barcharts?</div><b>To detect class imbalance</b>

A logistic regression model compute the following linear score for a student.<br>z=2<br>
if the classification threshhold is 0.5 determine the predicted class(class 0 if z<0.5 otherwise class 01)<br>


```python
import numpy as np
def sigmoid_func(z):
    return 1/(1+np.exp(-z))

def det_class(sigma):
    return ('class 01' if sigma>=0.5 else 'class 0')

z=2
print(det_class(sigmoid_func(z))) 
```
